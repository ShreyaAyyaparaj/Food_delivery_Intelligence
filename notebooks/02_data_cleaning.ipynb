```python
import pandas as pd
import json
with open("../data/raw/India-Food-Delivery-Time-Prediction.txt", "r", encoding="utf-8") as file:
    record = json.load(file)

df = pd.DataFrame(record)
df.head()
```

```python
categorical_columns = [
    "Weatherconditions",
    "Road_traffic_density",
    "Festival",
    "City",
    "Type_of_order",
    "Type_of_vehicle"
]
for column in categorical_columns:
    print(f"\n---{column}---")
    print(df[column].value_counts(dropna=False))
```

```python
for column in [
    "City",
    "Type_of_order",
    "Type_of_vehicle"
]:
    print(f"\n---{column}---")
    print(df[column].value_counts(dropna=False))
```

```python
# Clean Weatherconditions
df["Weatherconditions"] = (
    df["Weatherconditions"]
    .str.replace("conditions ", "", regex=False)
    .str.strip()
)

# Remove whitespace from categorical columns
categorical_columns = [
    "Weatherconditions",
    "Road_traffic_density",
    "Festival",
    "City",
    "Type_of_order",
    "Type_of_vehicle"
]

for column in categorical_columns:
    df[column] = df[column].str.strip()
```

```python
import pandas as pd
from pathlib import Path

# Convert time strings that contain the default date into time objects
for column in ["Time_Orderd", "Time_Order_picked"]:
    parsed = pd.to_datetime(df[column], errors="coerce")
    df[column] = parsed.dt.time

# Clean delivery duration and convert it to numeric
df["Time_taken(min)"] = pd.to_numeric(
    df["Time_taken(min)"]
    .astype(str)
    .str.replace("(min)", "", regex=False)
    .str.strip(),
    errors="coerce"
)

# Ensure all numeric columns are numeric
numeric_columns = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries",
    "Time_taken(min)"
]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

# Rebuild full timestamps from the date and time
df["order_datetime"] = pd.to_datetime(
    df["Order_Date"].dt.strftime("%Y-%m-%d") + " " +
    df["Time_Orderd"].astype(str),
    errors="coerce"
)

df["pickup_datetime"] = pd.to_datetime(
    df["Order_Date"].dt.strftime("%Y-%m-%d") + " " +
    df["Time_Order_picked"].astype(str),
    errors="coerce"
)

# Save corrected CSV
output_path = Path("../data/processed/food_delivery_clean.csv")
output_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(output_path, index=False)

print("Saved:", output_path.resolve())
print("Shape:", df.shape)
print(df[[
    "Time_Orderd",
    "Time_Order_picked",
    "Time_taken(min)",
    "order_datetime",
    "pickup_datetime"
]].head())
print(df[["Time_Orderd", "Time_taken(min)"]].isna().sum())
print(df[["Time_Orderd", "Time_taken(min)"]].dtypes)
```
